```python
from qiskit.quantum_info import *
from qiskit . visualization import plot_histogram
from numpy import sqrt
from IPython.display import Math, display

# Pauli operators

X = Operator([[0, 1],
              [1, 0]])

Y = Operator([[0, -1.0j],
              [1.0j, 0]])

Z = Operator([[1, 0],
              [0, -1]])

# Single-qubit gates

H = Operator([[1/sqrt(2), 1/sqrt(2)],
              [1/sqrt(2), -1/sqrt(2)]])

S = Operator([[1, 0],
              [0, 1.0j]])

T = Operator([[1, 0],
              [0, (1 + 1.0j)/sqrt(2)]])

# initial state |0>

v = Statevector([1, 0])

# evolve through sequence of operators

v = v.evolve(H)
display(v.draw("latex"))
v = v.evolve(T)
display(v.draw("latex"))
v = v.evolve(H)
display(v.draw("latex"))
v = v.evolve(T)
display(v.draw("latex"))
v = v.evolve(Z)
display(v.draw("latex"))

# ============================================================
# Operator(...) in Qiskit — What it does
# ============================================================

# Operator is used to represent a quantum operation (unitary or general)
# as a MATRIX (linear operator acting on statevectors).

# ------------------------------------------------------------
# 1. Creating an Operator
# ------------------------------------------------------------

# You can create it from:
# - a matrix (list or numpy array)
# - a QuantumCircuit
# - another operator-like object

# Example:
# Operator([[1,0],[0,1]]) → Identity operator


# ------------------------------------------------------------
# 2. From Circuit → Matrix
# ------------------------------------------------------------

# Operator(qc)

# This converts the ENTIRE quantum circuit into its unitary matrix.

# Example:
# Operator(qc_cx) → gives 4×4 matrix of CX gate

# Important:
# - Order is in computational basis: |00>, |01>, |10>, |11>


# ------------------------------------------------------------
# 3. Acting on states
# ------------------------------------------------------------

# If U = Operator(...)
# and |ψ> is a statevector:

# |ψ'> = U |ψ>

# In Qiskit:
# sv_out = sv_in.evolve(U)


# ------------------------------------------------------------
# 4. Drawing the matrix
# ------------------------------------------------------------

# Operator(qc).draw("latex")

# Displays the matrix in clean LaTeX form

# Useful for:
# - verifying identities (like CX, CZ, SWAP)
# - seeing exact transformation


# ------------------------------------------------------------
# 5. Composition / multiplication
# ------------------------------------------------------------

# Operators behave like matrices:

# U_total = U2 @ U1

# Meaning:
# first apply U1, then U2

# (same rule as matrix multiplication)


# ------------------------------------------------------------
# 6. Why it's useful
# ------------------------------------------------------------

# - Converts circuits → exact math form
# - Helps check equivalence of circuits
# - Bridges "circuit picture" ↔ "linear algebra picture"


# ------------------------------------------------------------
# ONE-LINE SUMMARY
# ------------------------------------------------------------

# Operator(...) = "give me the matrix form of this quantum operation"
```
