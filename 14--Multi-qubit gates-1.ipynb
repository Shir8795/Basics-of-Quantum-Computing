```python
from qiskit import QuantumRegister, QuantumCircuit
from numpy import pi
from qiskit.quantum_info import *
from qiskit . visualization import plot_histogram
from numpy import sqrt
from IPython.display import Math, display

# ============================================================
# Controlled Gates (2-qubit gates)
# ============================================================

# General idea:
# A controlled gate applies an operation on target qubit
# ONLY IF control qubit = |1>

# Notation:
# |control, target>

# ------------------------------------------------------------
# 1. CX (Controlled-X / CNOT)
# ------------------------------------------------------------
# If control = 1 → apply X (flip) on target
# If control = 0 → do nothing
#
# Action:
# |00> → |00>
# |01> → |01>
# |10> → |11>
# |11> → |10>
#
# Effect:
# - Creates entanglement
# - Most important 2-qubit gate

# ------------------------------------------------------------
# 2. CY (Controlled-Y)
# ------------------------------------------------------------
# If control = 1 → apply Y on target
#
# Y introduces phase (i), so:
# - flips + adds complex phase

# ------------------------------------------------------------
# 3. CZ (Controlled-Z)
# ------------------------------------------------------------
# If control = 1 → apply Z on target
#
# Effect:
# - does NOT change amplitudes
# - only adds phase to |11>
#
# |11> → -|11>

# ------------------------------------------------------------
# 4. CH (Controlled-Hadamard)
# ------------------------------------------------------------
# If control = 1 → apply H on target
#
# Effect:
# - creates superposition conditionally

# =======================
# Define 2-qubit register
# =======================
qreg = QuantumRegister(2, 'q')

# =======================
# Initial state |10>
# =======================
# control = 1, target = 0
sv_in = Statevector.from_label("11")

print("Initial State:\n")
display(sv_in.draw("latex"))

# ============================================================
# 1. CX Gate
# ============================================================
qc_cx = QuantumCircuit(qreg)
qc_cx.cx(qreg[0], qreg[1])

sv_out_cx = sv_in.evolve(qc_cx)

print("\nCX Gate:\n")
display(qc_cx.draw(output='mpl'))

print("Output:\n")
display(sv_out_cx.draw("latex"))

# ============================================================
# 2. CY Gate
# ============================================================
qc_cy = QuantumCircuit(qreg)
qc_cy.cy(qreg[0], qreg[1])

sv_out_cy = sv_in.evolve(qc_cy)

print("\nCY Gate:\n")
display(qc_cy.draw(output='mpl'))

print("Output:\n")
display(sv_out_cy.draw("latex"))

# ============================================================
# 3. CZ Gate
# ============================================================
qc_cz = QuantumCircuit(qreg)
qc_cz.cz(qreg[0], qreg[1])

sv_out_cz = sv_in.evolve(qc_cz)

print("\nCZ Gate:\n")
display(qc_cz.draw(output='mpl'))

print("Output:\n")
display(sv_out_cz.draw("latex"))

# ============================================================
# 4. CH Gate
# ============================================================
qc_ch = QuantumCircuit(qreg)
qc_ch.ch(qreg[0], qreg[1])

sv_out_ch = sv_in.evolve(qc_ch)

print("\nCH Gate:\n")
display(qc_ch.draw(output='mpl'))

print("Output:\n")
display(sv_out_ch.draw("latex"))

# qubit ordering is LITTLE ENDIAN

print("\nMatrix Forms (LaTeX):\n")

# CX
CX_op = Operator(qc_cx)
print("CX Matrix:\n")
display(CX_op.draw("latex"))

# CY
CY_op = Operator(qc_cy)
print("CY Matrix:\n")
display(CY_op.draw("latex"))

# CZ
CZ_op = Operator(qc_cz)
print("CZ Matrix:\n")
display(CZ_op.draw("latex"))

# CH
CH_op = Operator(qc_ch)
print("CH Matrix:\n")
display(CH_op.draw("latex"))
```
