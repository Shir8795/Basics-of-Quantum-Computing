```python
# ============================================================
# U3 Gate (General Single-Qubit Gate)
# ============================================================

# Any single-qubit unitary (up to global phase) can be written as:

# U3(θ, φ, λ) =
# [[ cos(θ/2),           -e^{iλ} sin(θ/2) ],
#  [ e^{iφ} sin(θ/2),    e^{i(φ+λ)} cos(θ/2) ]]

# Meaning of parameters:
# θ → controls mixing between |0> and |1> (amplitude change)
# φ → phase added after mixing
# λ → phase before mixing

# Very important decomposition:
# U3(θ, φ, λ) = Rz(φ) · Ry(θ) · Rz(λ)

# So physically:
# 1. Rotate around Z by λ
# 2. Rotate around Y by θ
# 3. Rotate around Z by φ

# This shows:
# → Any single-qubit gate = combination of Rz and Ry
# → U3 is the most general SU(2) rotation

# Special cases:
# Rx(θ) = U3(θ, -π/2, π/2)
# Ry(θ) = U3(θ, 0, 0)
# Rz(θ) = U3(0, 0, θ)

from qiskit.visualization import plot_bloch_multivector
from qiskit import QuantumRegister, QuantumCircuit
from numpy import pi, cos, sin, exp
from qiskit.quantum_info import Operator, Statevector
from IPython.display import display

# =======================
# Parameters
# =======================
theta = pi/2
phi = pi/3
lam = pi/4

# =======================
# Define U3 matrix manually
# =======================
U3_op = Operator([
    [cos(theta/2), -exp(1j*lam)*sin(theta/2)],
    [exp(1j*phi)*sin(theta/2), exp(1j*(phi+lam))*cos(theta/2)]
])

# =======================
# Circuit
# =======================
qreg = QuantumRegister(1, 'q')
qc = QuantumCircuit(qreg)

# Apply U3 gate (Qiskit built-in)
qc.u(theta, phi, lam, qreg[0])

print("U3 Gate Matrix:\n")
display(U3_op)

print("Circuit:\n")
display(qc.draw(output='mpl'))

# =======================
# State evolution
# =======================
sv_in = Statevector.from_label("+")
sv_out = sv_in.evolve(qc)

print("Input State:\n")
display(sv_in.draw("bloch"))
display(sv_in.draw("latex"))

print("Output State:\n")
display(sv_out.draw("bloch"))
display(sv_out.draw("latex"))
```
