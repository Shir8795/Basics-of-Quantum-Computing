```python
# Unitary decomposition and universal quantum gates
#
# 1) Decomposition of a d-dimensional unitary into 2-level unitaries
# Consider a d-dimensional Hilbert space. For d = 3,
# U = [[a,d,g],[b,e,h],[c,f,j]].
# Goal: find 2-level unitaries U1,U2,U3,... such that Uk...U2 U1 U = I,
# hence U = U1† U2† ... Uk†.
#
# A 2-level unitary acts non-trivially only on a pair of basis states.
# Example, acting on |1>,|2> and leaving |3> unchanged:
# U2lev = [[a',b',0],[c',d',0],[0,0,1]].
# Unitarity requires |a'|²+|b'|²=1, |c'|²+|d'|²=1,
# and a'*c' + b'*d' = 0.
#
# For the 3×3 case, choose U1 to eliminate b in the first column:
# c'a + d'b = 0  =>  c'/d' = -b/a.
# Let N = sqrt(|a|²+|b|²). One choice is
# U1 = [[a*/N,b*/N,0],[b/N,-a/N,0],[0,0,1]].
# Then U1 U = [[a1,d1,g1],[0,e1,h1],[c1,f1,j1]].
#
# Next choose U2 acting on |1>,|3> to eliminate c1.
# Let N1 = sqrt(|a1|²+|c1|²):
# U2 = [[a1*/N1,0,c1*/N1],[0,1,0],[c1/N1,0,-a1/N1]].
# Thus U2 U1 U = [[a2,d2,g2],[0,e2,h2],[0,f2,j2]].
#
# Since U2 U1 U is unitary, |a2|=1 and orthogonality forces d2=g2=0.
# The remaining lower 2×2 block can be removed by another 2-level unitary U3.
# Therefore U can be written as a product of 2-level unitaries.
#
# In d dimensions, the first column needs at most (d-1) eliminations,
# the second needs (d-2), ..., the last needs 1:
# k <= (d-1)+(d-2)+...+1 = d(d-1)/2.
#
# For an n-qubit system, d = 2^n:
# number of 2-level unitaries <= 2^(n-1)(2^n-1) = O(4^n).
#
# 2) Implementing a 2-level unitary with CNOTs + 1-qubit gates
# Any 2-level unitary mixes only two computational-basis states.
# Example: U|000> = a|000> + b|111>, with the corresponding 2×2 block
# U~ = [[a,c],[b,d]].
#
# Use CNOT/X-type shuffling to map the two selected basis states to a pair
# that differs in only one qubit, apply a suitably controlled U~ on that qubit,
# then undo the shuffling.
# A control-on-0 can be converted to control-on-1 by placing X gates
# before and after that control line.
#
# The shuffle/unshuffle requires O(n) multi-controlled operations.
# Each such operation can be reduced using O(n) CNOT/single-qubit operations.
# Hence one 2-level unitary can be implemented with about O(n²)
# CNOT + single-qubit gates.
#
# 3) Controlled-U
# C^m(U)|x1,x2,...,xm>|psi> applies U to |psi> only when all controls xi=1.
#
# A singly controlled U can be decomposed using
# U = exp(i alpha) A X B X C, with ABC = I.
# Circuit form: C -- CNOT -- B -- CNOT -- A on the target,
# together with C and the phase diag(1,e^(i alpha)) as in the notes.
# Therefore a controlled 1-qubit U needs only CNOTs + 1-qubit gates.
#
# For a doubly controlled U, choose V such that V² = U.
# Then C²(U) can be decomposed into controlled-V, controlled-V†,
# controlled-V and CNOTs between the control lines.
# The same idea can be applied recursively for more controls.
#
# 4) Toffoli and multi-controlled gates
# A Toffoli gate can be decomposed using {CNOT, H, T, S}, with S = T².
# A many-controlled U can be built using ancillas:
# compute the logical AND of the controls with a chain of Toffolis,
# apply the required controlled-U on the target, then uncompute the ancillas.
#
# Example structure for C^5(U):
# 4 Toffolis to compute the control condition,
# one controlled-U on the target,
# then 4 Toffolis to uncompute.
#
# Thus a general multi-controlled operation reduces to CNOTs + 1-qubit gates.
# Consequently {CNOT, arbitrary 1-qubit gates} is universal.
#
# 5) From arbitrary 1-qubit gates to {H,T}
# Since S = T², the set {CNOT,H,T} is sufficient as a universal set.
# H and T can generate/approximate the required 1-qubit rotations.
#
# X = H Z H = H T^4 H.
# T = [[1,0],[0,e^(i pi/4)]] corresponds to a z-axis rotation up to phase.
# H T H gives the corresponding x-axis rotation.
#
# Because x and z are perpendicular axes, a general 1-qubit unitary can be written
# in Euler form:
# U = exp(i alpha) Rz(beta) Rx(gamma) Rz(delta).
# The angles can be obtained by matching this form to the given 2×2 unitary.
#
# Arbitrary rotations can be approximated by finite sequences of H and T.
# By the Solovay-Kitaev theorem, an approximation with error epsilon uses
# O(log^c(1/epsilon)) gates, with c approximately 2 as used in these notes.
#
# 6) Overall scaling
# Any n-qubit unitary:
#   U -> O(4^n) 2-level unitaries
#      -> each 2-level unitary -> O(n²) CNOT + 1-qubit gates
#      -> each 1-qubit unitary -> an H,T sequence via Solovay-Kitaev.
#
# Hence the total {CNOT,H,T} gate count is written schematically as
# O(n² 4^n log^c(n² 4^n / epsilon)),
# approximately O(n^(2+c) 4^n log^c(1/epsilon)).
```
