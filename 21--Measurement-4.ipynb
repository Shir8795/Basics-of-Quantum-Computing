```python
from qiskit.quantum_info import Statevector, DensityMatrix, partial_trace
from IPython.display import display
from numpy import sqrt

# ============================================================
# DEFINE 2-QUBIT STATE
# ============================================================

# Example: Bell state
# |ψ> = (|00> + |11>) / √2

psi2 = Statevector([1/sqrt(2), 0, 0, 1/sqrt(2)])

print("2-Qubit State |ψ> :")
display(psi2.draw("latex"))

# ============================================================
# DENSITY MATRIX
# ============================================================

rho2 = DensityMatrix(psi2)

print("\nDensity Matrix (2-qubit):")
display(rho2.draw("latex"))

# ============================================================
# PARTIAL TRACE
# ============================================================

# Trace out qubit 1 → keep qubit 0
rho_reduced_0 = partial_trace(rho2, [1])

print("\nReduced Density Matrix (keeping qubit 0):")
display(rho_reduced_0.draw("latex"))

# Trace out qubit 0 → keep qubit 1
rho_reduced_1 = partial_trace(rho2, [0])

print("\nReduced Density Matrix (keeping qubit 1):")
display(rho_reduced_1.draw("latex"))
```
