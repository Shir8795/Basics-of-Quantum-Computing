```python
from qiskit.quantum_info import Statevector
from numpy import sqrt

# ============================================================
# DEFINE STATES
# ============================================================

# |ψ> = |0> = [1, 0]
psi = Statevector([1, 0])

# |ψ'> = |+> = (|0> + |1>) / √2 = [1/√2, 1/√2]
psi_prime = Statevector([1/sqrt(2), 1/sqrt(2)])


# ============================================================
# SHOW STATES
# ============================================================

# This prints the states in Dirac notation
print("State |ψ> :")
display(psi.draw("latex"))

print("\nState |ψ'> :")
display(psi_prime.draw("latex"))

# ============================================================
# INNER PRODUCT
# ============================================================

# psi_prime.data gives array form of |ψ'>
# psi.data gives array form of |ψ>

# psi_prime.data.conj() → gives ⟨ψ'| (bra vector)
# @ → matrix multiplication

overlap = psi_prime.data.conj() @ psi.data


# ============================================================
# WHAT THIS LINE DOES (VERY IMPORTANT)
# ============================================================

# Mathematically:
# ⟨ψ'|ψ⟩ = sum over i of (ψ'_i)* × ψ_i

# For our case:
# ψ' = [1/√2, 1/√2]
# ψ  = [1, 0]

# So:
# ⟨ψ'|ψ⟩ = (1/√2)* × 1 + (1/√2)* × 0
#         = 1/√2

print("\nInner product ⟨ψ'|ψ⟩ = ", overlap)


# ============================================================
# PROBABILITY
# ============================================================

# Probability = |⟨ψ'|ψ⟩|²
P = abs(overlap)**2

print("\nProbability = ", P)
```
